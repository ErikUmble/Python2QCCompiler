### Plot by Num Vars and Complexity

```python
def get_probability_data(db, compiler, function_success_threshold, num_vars_iter = range(2, 21), complexity_iter = range(1, 21)):
    """
    function_success_threshold: the minimum success rate for a function to be considered "success" by the compiler
    if function_success_threshold is None, this returns the mean success rate (fidelity) across all functions for each (num_vars, complexity) pair; 
    otherwise, it returns the probability that a random function with that (num_vars, complexity) is successfully compiled by the compiler
    """
    num_vars_data = []
    complexity_data = []
    probability_data = []
    mean_success_data = []

    for num_vars in num_vars_iter:
        for complexity in complexity_iter:
            print(f"(num_vars, complexity) = ({num_vars}, {complexity})")
            mean_success_rate = 0.0
            successful_function_count = 0
            function_count = 0
            no_data = True
            problem_instances = db.find_problem_instances(num_vars=num_vars, complexity=complexity)
            #problem_instances = db.find_problem_instances(size_filters={"num_vars": num_vars, "complexity": complexity})
            trials_by_function = {instance.id: db.find_trials(instance_id=instance.id, compiler_name=compiler.name) for instance in problem_instances}
            #trials_by_function = trials.get_per_statement(num_vars=num_vars, complexity=complexity)
            for function in trials_by_function.keys():
                # skip functions with no trials for this compiler
                if len(trials_by_function[function]) == 0:
                    continue
                function_count += 1
                no_data = False
                
                s = 0.0
                for trial in trials_by_function[function]:
                    s += trial.calculate_success_rate()
                    
                if function_success_threshold is not None and s / len(trials_by_function[function]) > function_success_threshold:
                    successful_function_count += 1
                mean_success_rate += s / len(trials_by_function[function])

            if no_data:
                 print(f"Warning: no results for {num_vars} variables, complexity {complexity}; skipping")
                 continue
            
            num_vars_data.append(num_vars)
            complexity_data.append(complexity)
            probability_data.append(successful_function_count / function_count if function_count > 0 else 0)
            mean_success_data.append(mean_success_rate / function_count if function_count > 0 else 0)

    if function_success_threshold is None:
        return num_vars_data, complexity_data, mean_success_data
    
    return num_vars_data, complexity_data, probability_data


def plot_probability_data(num_vars_data, complexity_data, probability_data, title, filepath=None):
    
    df = pd.DataFrame({
        'num_vars': num_vars_data,
        'complexity': complexity_data,
        'prob': probability_data
    })
    grid = df.pivot(index='complexity', columns='num_vars', values='prob')
    grid = grid.sort_index(ascending=True)

    plt.figure(figsize=(12, 10))
    plt.pcolormesh(
        grid.columns, grid.index, grid.values,
        cmap='RdYlGn', shading='auto'
    )

    plt.xticks(np.arange(min(num_vars_data), max(num_vars_data) + 1, 1))
    plt.yticks(np.arange(min(complexity_data), max(complexity_data) + 1, 1))

    plt.xlabel('Variables Count')
    plt.ylabel('Complexity')
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(f'Mean Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()

def plot_probability_data_side_by_side(num_vars_data, complexity_data, probability_data1, probability_data2, title1, title2, filepath=None):
    df1 = pd.DataFrame({
        'num_vars': num_vars_data,
        'complexity': complexity_data,
        'prob': probability_data1
    })
    grid1 = df1.pivot(index='complexity', columns='num_vars', values='prob')
    grid1 = grid1.sort_index(ascending=True)

    df2 = pd.DataFrame({
        'num_vars': num_vars_data,
        'complexity': complexity_data,
        'prob': probability_data2
    })
    grid2 = df2.pivot(index='complexity', columns='num_vars', values='prob')
    grid2 = grid2.sort_index(ascending=True)

    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(24, 10))
    plot1 = axes[0].pcolormesh(
        grid1.columns, grid1.index, grid1.values,
        cmap='RdYlGn', shading='auto'
    )
    axes[0].set_xticks(np.arange(min(num_vars_data), max(num_vars_data) + 1, 1))
    axes[0].set_yticks(np.arange(min(complexity_data), max(complexity_data) + 1, 1))
    axes[0].set_xlabel('Variables Count')
    axes[0].set_ylabel('Complexity')
    axes[0].set_title(title1)

    plot2 = axes[1].pcolormesh(
        grid2.columns, grid2.index, grid2.values,
        cmap='RdYlGn', shading='auto'
    )
    axes[1].set_xticks(np.arange(min(num_vars_data), max(num_vars_data) + 1, 1))
    axes[1].set_yticks(np.arange(min(complexity_data), max(complexity_data) + 1, 1))
    axes[1].set_xlabel('Variables Count')
    axes[1].set_ylabel('Complexity')
    axes[1].set_title(title2)

    cbar = fig.colorbar(plot1, ax=axes.ravel().tolist(), shrink=0.95)
    cbar.set_label(f'Mean Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()


def plot_probability_difference(num_vars_data, complexity_data, probability_data1, probability_data2, title, filepath=None):
    difference_data = [p1 - p2 for p1, p2 in zip(probability_data1, probability_data2)]
    df = pd.DataFrame({
        'num_vars': num_vars_data,
        'complexity': complexity_data,
        'diff': difference_data
    })
    grid = df.pivot(index='complexity', columns='num_vars', values='diff')
    grid = grid.sort_index(ascending=True)
    plt.figure(figsize=(12, 10))
    plt.pcolormesh(
        grid.columns, grid.index, grid.values,
        cmap='RdBu_r', shading='auto', vmin=-0.4, vmax=0.4
    )
    plt.xticks(np.arange(min(num_vars_data), max(num_vars_data) + 1, 1))
    plt.yticks(np.arange(min(complexity_data), max(complexity_data) + 1, 1))

    plt.xlabel('Variables Count')
    plt.ylabel('Complexity')
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(f'Difference in Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()
```

```python
def get_stddev_data(db, use_hamming=False):
    # compute standard deviations of exact match rate across different trials for the same function
    
    num_vars_data = []
    complexity_data = []
    all_stddev_data = {}

    for num_vars in range(2, 33): # 2 - 32
        for complexity in range(1, 22): # 1 - 21
            print(f"(num_vars, complexity) = ({num_vars}, {complexity})")
    
            stddev_data = []
            no_data = True
            problem_instances = db.find_problem_instances(size_filters={'num_vars': num_vars, 'complexity': complexity})
            trials_by_function = {instance.instance_id: db.find_trials(instance_id=instance.instance_id, compiler_name=compiler.name) for instance in problem_instances}
            #trials_by_function = trials.get_per_statement(num_vars=num_vars, complexity=complexity)

            for function in trials_by_function.keys():
                # skip functions with no trials for this compiler
                if len(trials_by_function[function]) == 0:
                    continue

                no_data = False
                
                accuracy_rates = []
                for trial in trials_by_function[function]:
                    if use_hamming:
                        accuracy_rates.append(1-trial.mean_hamming_distance)
                    else:
                        accuracy_rates.append(trial.result_qubit_success_rate)

                stddev_data.append(np.std(accuracy_rates))
                    
            if no_data:
                 print(f"Warning: no results for {num_vars} variables, complexity {complexity}; skipping")
                 continue
            
            num_vars_data.append(num_vars)
            complexity_data.append(complexity)
            all_stddev_data[(num_vars, complexity)] = stddev_data

    return all_stddev_data, num_vars_data, complexity_data


```

### Problem Difficulty Plots

```python
def get_fidelity_data(oracle_db, compiler_name=None, n_iter=None, num_bins=6):
    n_data = []
    num_solutions_data = []
    fidelity_data = []

    _aggregate_data = {}
    n_iter = list(n_iter) if n_iter is not None else list(range(1, 6))
    
    # split into batches based on n to avoid loading all trials into memory at once
    for n in n_iter:
        problems = oracle_db.query(
            select(RandomBooleanFunction)
            .where(RandomBooleanFunction.num_vars == n)
        )
        for problem in problems:
            trials = oracle_db.query(
                select(RandomBooleanFunctionTrial)
                .join(RandomBooleanFunctionTrial.problem)
                .where(RandomBooleanFunctionTrial.compiler_name == compiler_name)
                .where(RandomBooleanFunctionTrial.problem_id == problem.id)
                .options(joinedload(RandomBooleanFunctionTrial.problem))
            )
            if len(trials) == 0:
                continue
            
            problem_fidelities = []
            n = problem.number_of_input_bits()
            num_solutions = problem.get_num_solutions()
            for trial in trials:
                if trial.is_failed:
                    problem_fidelities.append(0.0)
                    continue
                problem_fidelities.append(trial.calculate_success_rate())

            problem_fidelity = np.mean(problem_fidelities)
            # form bins based on the number of solutions compared with total search space size 2^n
            # scale with n: thresholds are 2^(n-5), 2^(n-4), ..., 2^(n-1)
            bin_index = None
            if num_solutions == 0:
                bin_index = 0
            else:
                for i in range(num_bins - 1):
                    if num_solutions <= 2 ** (n - (num_bins - 1 - i)):
                        bin_index = i + 1
                        break
                if bin_index is None:
                    bin_index = num_bins - 1

           
            _aggregate_data[(n, bin_index)] = _aggregate_data.get((n, bin_index), []) + [problem_fidelity]

    for n in n_iter:
        for i in range(num_bins):
            if (n, i) not in _aggregate_data:
                _aggregate_data[(n, i)] = []

            n_data.append(n)
            num_solutions_data.append(i)
            fidelity_data.append(np.mean(_aggregate_data[(n, i)]))
    
    return n_data, num_solutions_data, fidelity_data

def _get_yticks(num_solutions_data, num_bins):
    low = min(num_solutions_data)
    high = max(num_solutions_data)
    indices = []
    labels = []
    # thresholds corresponding to bin upper bounds: 0, 2^(n-5), 2^(n-4), 2^(n-3), 2^(n-2), 2^(n-1)
    bin_labels = ['0']
    for i in range(num_bins - 1):
        bin_labels.append(f"2^(n-{num_bins -1 - i})")
    for i in range(low, high):
        indices.append(i + 0.5)
        labels.append(bin_labels[i])
    return np.array(indices), labels

def plot_fidelity_data(n_data, num_solutions_data, fidelity_data, title, filepath=None):

    df = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'fidelity': fidelity_data
    })
    grid = df.pivot(index='num_solutions', columns='num_vars', values='fidelity')
    grid = grid.sort_index(ascending=True)

    # set NaN values to white since we do not have a full rectangle of data
    cmap = plt.cm.get_cmap("RdYlGn").copy()
    cmap.set_bad(color="white")

    x_range = range(min(n_data), max(n_data) + 1)
    y_range = range(min(num_solutions_data), max(num_solutions_data) + 1)
    figsize=(x_range.stop - x_range.start, y_range.stop - y_range.start)
    plt.figure(figsize=figsize)
    plt.pcolormesh(
        grid.columns,
        grid.index,
        grid.values,
        cmap=cmap,
        shading='nearest',
        vmin=0,
        vmax=0.2
    )
    num_bins = max(num_solutions_data) + 1
    plt.xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    plt.yticks(*_get_yticks(num_solutions_data, num_bins))

    plt.xlabel('Number of Variables (n)')
    plt.ylabel('Number of Solutions (m)')
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(f'Mean Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()

def plot_fidelity_data_side_by_side(n_data, num_solutions_data, fidelity_data1, fidelity_data2, title1, title2, filepath=None):
    df1 = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'fidelity': fidelity_data1
    })
    grid1 = df1.pivot(index='num_solutions', columns='num_vars', values='fidelity')
    grid1 = grid1.sort_index(ascending=True)

    df2 = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'fidelity': fidelity_data2
    })
    grid2 = df2.pivot(index='num_solutions', columns='num_vars', values='fidelity')
    grid2 = grid2.sort_index(ascending=True)

    x_range = range(min(n_data), max(n_data) + 1)
    y_range = range(min(num_solutions_data), max(num_solutions_data) + 1)
    figsize=((x_range.stop - x_range.start)*3 //5, y_range.stop - y_range.start)
    fig, axes = plt.subplots(nrows=2, ncols=1, figsize=figsize)
    plot1 = axes[0].pcolormesh(
        grid1.columns, grid1.index, grid1.values,
        cmap='RdYlGn', shading='auto', vmin=0, vmax=0.2
    )
    num_bins = max(num_solutions_data) + 1
    axes[0].set_xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    axes[0].set_yticks(*_get_yticks(num_solutions_data, num_bins))
    axes[0].set_xlabel('Number of Variables (n)')
    axes[0].set_ylabel('Number of Solutions (m)')
    axes[0].set_title(title1)

    plot2 = axes[1].pcolormesh(
        grid2.columns, grid2.index, grid2.values,
        cmap='RdYlGn', shading='auto', vmin=0, vmax=0.2
    )
    axes[1].set_xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    axes[1].set_yticks(*_get_yticks(num_solutions_data, num_bins))
    axes[1].set_xlabel('Number of Variables (n)')
    axes[1].set_ylabel('Number of Solutions (m)')
    axes[1].set_title(title2)

    cbar = fig.colorbar(plot1, ax=axes.ravel().tolist(), shrink=0.95)
    cbar.set_label(f'Mean Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()

def _plot_fidelity_data_side_by_side_horizontal(n_data, num_solutions_data, fidelity_data1, fidelity_data2, title1, title2, filepath=None):
    df1 = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'fidelity': fidelity_data1
    })
    grid1 = df1.pivot(index='num_solutions', columns='num_vars', values='fidelity')
    grid1 = grid1.sort_index(ascending=True)

    df2 = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'fidelity': fidelity_data2
    })
    grid2 = df2.pivot(index='num_solutions', columns='num_vars', values='fidelity')
    grid2 = grid2.sort_index(ascending=True)

    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 8))
    plot1 = axes[0].pcolormesh(
        grid1.columns, grid1.index, grid1.values,
        cmap='RdYlGn', shading='auto', vmin=0, vmax=0.2
    )
    num_bins = max(num_solutions_data) + 1
    axes[0].set_xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    axes[0].set_yticks(*_get_yticks(num_solutions_data, num_bins))
    axes[0].set_xlabel('Number of Variables (n)')
    axes[0].set_ylabel('Number of Solutions (m)')
    axes[0].set_title(title1)

    plot2 = axes[1].pcolormesh(
        grid2.columns, grid2.index, grid2.values,
        cmap='RdYlGn', shading='auto', vmin=0, vmax=0.2
    )
    axes[1].set_xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    axes[1].set_yticks(*_get_yticks(num_solutions_data, num_bins))
    axes[1].set_xlabel('Number of Variables (n)')
    axes[1].set_ylabel('Number of Solutions (m)')
    axes[1].set_title(title2)

    cbar = fig.colorbar(plot1, ax=axes.ravel().tolist(), shrink=0.95)
    cbar.set_label(f'Mean Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()


def plot_fidelity_difference(n_data, num_solutions_data, fidelity_data1, fidelity_data2, title, filepath=None):
    difference_data = [f1 - f2 for f1, f2 in zip(fidelity_data1, fidelity_data2)]
    df = pd.DataFrame({
        'num_vars': n_data,
        'num_solutions': num_solutions_data,
        'diff': difference_data
    })
    grid = df.pivot(index='num_solutions', columns='num_vars', values='diff')
    grid = grid.sort_index(ascending=True)

    x_range = range(min(n_data), max(n_data) + 1)
    y_range = range(min(num_solutions_data), max(num_solutions_data) + 1)
    figsize=(x_range.stop - x_range.start, y_range.stop - y_range.start)
    plt.figure(figsize=figsize)
    plt.pcolormesh(
        grid.columns, grid.index, grid.values,
        cmap='RdBu_r', shading='auto', vmin=-0.2, vmax=0.2
    )
    num_bins = max(num_solutions_data) + 1
    plt.xticks(np.arange(min(n_data), max(n_data) + 1, 1))
    plt.yticks(*_get_yticks(num_solutions_data, num_bins))

    plt.xlabel('Number of Variables (n)')
    plt.ylabel('Number of Solutions (m)')
    plt.title(title)
    cbar = plt.colorbar()
    cbar.set_label(f'Difference in Fidelity', rotation=270, labelpad=15)
    if filepath is not None:
        plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.show()
```
